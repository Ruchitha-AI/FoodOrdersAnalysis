```python
# Create rating ranges
df['rating_range'] = pd.cut(
    df['rating'],
    bins=[0, 2, 3, 4, 5],
    labels=['0–2', '2–3', '3–4', '4–5']
)

df[['rating', 'rating_range']].head()
```

```python
rating_revenue = df.groupby('rating_range')['total_amount'].sum()
rating_revenue
```
